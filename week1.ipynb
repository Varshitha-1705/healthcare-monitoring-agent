```python
import pandas as pd
import sqlite3
from datetime import datetime

# Create a small sample health dataset
data = {
    "metric": ["heart_rate", "steps", "sleep_hours", "weight"],
    "value": [78, 5234, 7.2, 58],
    "recorded_at": [
        "2025-10-19 09:00",
        "2025-10-19 09:00",
        "2025-10-18 23:00",
        "2025-10-18 08:00"
    ]
}

df = pd.DataFrame(data)
df.to_csv("sample_health_data.csv", index=False)
print("✅ sample_health_data.csv created")
df
```

```python
# Connect to a local SQLite database (it will create one if not present)
conn = sqlite3.connect("health_data.db")
c = conn.cursor()

# Create tables
c.execute("""CREATE TABLE IF NOT EXISTS medications (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT,
                dose TEXT,
                schedule TEXT,
                added_at TEXT
            )""")

c.execute("""CREATE TABLE IF NOT EXISTS metrics (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                metric TEXT,
                value TEXT,
                recorded_at TEXT
            )""")

conn.commit()
print("✅ Database and tables created successfully!")
```

```python
# Read the dataset
df = pd.read_csv("sample_health_data.csv")

# Insert data into metrics table
for _, row in df.iterrows():
    c.execute("INSERT INTO metrics (metric, value, recorded_at) VALUES (?, ?, ?)",
              (row["metric"], str(row["value"]), row["recorded_at"]))
conn.commit()

print("✅ Sample dataset inserted into metrics table!")
```

```python
name = input("Enter medication name: ")
dose = input("Enter dose (e.g., 500 mg): ")
schedule = input("Enter schedule (e.g., 08:00 daily): ")

c.execute("INSERT INTO medications (name, dose, schedule, added_at) VALUES (?, ?, ?, ?)",
          (name, dose, schedule, datetime.utcnow().isoformat()))
conn.commit()

print("✅ Medication added successfully!")
```

```python
print("\n💊 Medications:")
for row in c.execute("SELECT * FROM medications"):
    print(row)

print("\n📊 Health Metrics:")
for row in c.execute("SELECT * FROM metrics"):
    print(row)
```

```python
conn.close()
print("🔒 Database connection closed.")
```
