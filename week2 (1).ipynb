```python
import gradio as gr
import requests

# 👉 Replace this with your actual Groq API key from https://console.groq.com/keys
import os
os.environ["GROQ_API_KEY"] = "paste_your_new_key_here"
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

# Groq API endpoint and model
GROQ_URL = "https://api.groq.com/openai/v1/chat/completions"
GROQ_MODEL = "llama-3.3-70b-versatile"
```

```python
def get_ai_summary(heart_rate, steps, sleep_hours, weight):
    """
    Sends user health data to Groq LLaMA-3 model and returns a short summary.
    """
    prompt = (
        f"Patient metrics — Heart rate: {heart_rate} bpm, "
        f"Steps: {steps}, Sleep: {sleep_hours} hrs, "
        f"Weight: {weight} kg. "
        "Provide a short health summary and one daily advice."
    )

    try:
        headers = {
            "Authorization": f"Bearer {GROQ_API_KEY}",
            "Content-Type": "application/json"
        }
        payload = {
            "model": GROQ_MODEL,
            "messages": [{"role": "user", "content": prompt}]
        }
        res = requests.post(GROQ_URL, headers=headers, json=payload, timeout=30)

        # Handle API issues safely
        if res.status_code != 200:
            return f"⚠️ API Error: {res.status_code} – {res.text}"

        data = res.json()
        return data["choices"][0]["message"]["content"].strip()

    except Exception as e:
        return f"❌ Error: {e}"
```

```python
demo = gr.Interface(
    fn=get_ai_summary,
    inputs=[
        gr.Number(label="Heart Rate (bpm)", value=75),
        gr.Number(label="Steps", value=5000),
        gr.Number(label="Sleep Hours", value=7.0),
        gr.Number(label="Weight (kg)", value=60)
    ],
    outputs="text",
    title="🩺 Healthcare Monitoring AI Agent",
    description="Enter health data → Get AI-generated summary and daily advice using Groq LLaMA-3 model."
)

demo.launch(share=True)
```
